## 1. Domaine Métier

**Domaine :** Commerce électronique (E-commerce)  

Le e-commerce génère quotidiennement un volume massif de données : ventes, comportements d'achat, avis clients, délais de livraison, compositions de paniers. Ces données, correctement exploitées, permettent de prendre des décisions stratégiques précises et mesurables.

**Sources de données utilisées :**

| Dataset | Source | Volume | Utilisation |
|---|---|---|---|
| Olist Brazilian E-Commerce | Kaggle | ~100 000 commandes | Ventes, clients, paiements, livraisons |
| Instacart Market Basket | Kaggle | ~3M commandes, 200K clients | Habitudes d'achat, composition panier |
| Amazon Product Reviews | Kaggle | Avis & notes produits | Satisfaction client |

## 2. Problématique Métier

Dans un marché e-commerce fortement concurrentiel, **acquérir un nouveau client coûte 5 à 7 fois plus cher que de conserver un client existant** (source : Harvard Business Review).  

La plateforme souhaite donc répondre à la question stratégique suivante :

> **Comment exploiter les données de ventes, de comportement d'achat et de satisfaction client afin d'améliorer la fidélisation des clients et d'augmenter le chiffre d'affaires ?**

### Sous-questions décisionnelles

1. Quels clients présentent un risque élevé de désengagement ?
2. Quels facteurs influencent le plus la satisfaction client ?
3. Quels produits génèrent le plus de revenus ?
4. Quel impact les retards de livraison ont-ils sur la satisfaction ?
5. Quels segments de clients doivent être ciblés en priorité ?

## 3. KPI Principal

### Customer Retention Rate (Taux de Fidélisation)

Mesure la capacité de la plateforme à conserver ses clients actifs dans le temps.

$$\text{Taux de Fidélisation} = \frac{\text{Clients ayant effectué un nouvel achat}}{\text{Nombre total de clients}} \times 100$$

**Justification :**
- Les clients fidèles achètent plus fréquemment.
- Leur Lifetime Value (LTV) est significativement plus élevée.
- Le coût de rétention est inférieur au coût d'acquisition.

## 4. KPIs Secondaires

| Catégorie | KPI | Formule / Description |
|---|---|---|
| **Performance Commerciale** | Chiffre d'affaires total | Somme de tous les paiements |
| | Croissance des ventes | (CA mois N - CA mois N-1) / CA mois N-1 |
| | Nombre total de commandes | Count(order_id) |
| | Panier moyen | CA total / Nombre de commandes |
| **Activité Client** | Nombre de clients actifs | Clients ayant commandé dans les 90 derniers jours |
| | Taux de réachat | Clients avec ≥ 2 commandes / total clients |
| | Fréquence d'achat | Nombre de commandes / Nombre de clients |
| **Satisfaction Client** | Note moyenne des avis | Moyenne des review_scores |
| | % Avis positifs | Avis ≥ 4 étoiles / Total avis |
| | % Avis négatifs | Avis ≤ 2 étoiles / Total avis |
| **Logistique** | Délai moyen de livraison | Moyenne(date_livraison - date_commande) |
| | Taux de retard | Livraisons après date estimée / Total livraisons |
| **Produits** | Produits les plus vendus | Classement par quantité vendue |
| | Produits les plus rentables | Classement par CA généré |

## 5. KPI Tree

```
OBJECTIF STRATÉGIQUE : Augmenter le Chiffre d'Affaires
│
├── [A] Améliorer la Fidélisation Client  (KPI : Customer Retention Rate)
│   │
│   ├── [A1] Améliorer la Satisfaction Client
│   │   ├── Note moyenne des avis
│   │   ├── % Avis positifs
│   │   └── % Avis négatifs
│   │
│   ├── [A2] Améliorer la Qualité de Livraison
│   │   ├── Délai moyen de livraison
│   │   └── Taux de retard de livraison
│   │
│   └── [A3] Augmenter la Fréquence d'Achat
│       ├── Nombre de commandes par client
│       ├── Taux de réachat
│       └── Nombre de clients actifs
│
├── [B] Augmenter le Nombre de Commandes
│   ├── Produits les plus populaires
│   └── Acquisition et activation clients
│
└── [C] Augmenter le Panier Moyen
    ├── Taille moyenne du panier (nb articles)
    ├── Stratégie de cross-selling
    └── Promotion des produits premium
```

## 6. Business Case & ROI Attendu

### Hypothèse de base (à partir des données Olist)

| Indicateur | Valeur estimée |
|---|---|
| Nombre total de clients | ~100 000 |
| Taux de fidélisation actuel (estimé) | ~3–5% (e-commerce brésilien typique) |
| Panier moyen | ~120 BRL |
| Fréquence d'achat annuelle (clients fidèles) | ~3 commandes/an |

### Estimation du gain

Si le système permet d'augmenter le taux de fidélisation de **+5 points** (ex. de 5% à 10%) :

- Nouveaux clients fidélisés : 5 000 clients supplémentaires
- Gain CA additionnel estimé : 5 000 × 120 BRL × 3 = **1 800 000 BRL/an**

### ROI de la démarche Data

La démarche data (développement + infrastructure) est estimée à ~50 000 BRL.  
ROI estimé = (1 800 000 - 50 000) / 50 000 = **~3 500%**

> Ces chiffres sont des estimations conservatrices basées sur la littérature e-commerce. Ils seront affinés après l'EDA (Phase 4).
